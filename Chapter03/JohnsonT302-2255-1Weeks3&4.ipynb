*Exercise 3.05: Generating Arrays Using Arange and Linspace Methods*

```python
#np.arange() creates an array of evenly spaced values within a given interval.
#syntax np.arange(start, stop, interval count) if interval count not given goes by 1
np.arange(5,16)
```

```python
print("Numbers spaced apart by 2: ", np.arange(0,11,2))
print("Numbers spaced apart by a floating point number: ", np.arange(0,11,2.5))
print("Every 5th number from 30 in reverse order: ", np.arange(30, -1, -5))
```

```python
#np.linspace() creates an array of evenly spaced values within a specified range
#syntax - np.linspace(start, stop, num, endpoint, retstep, dtype)
#num - The number of samples to generate (default is 50)
print("11 linearly spaced numbers between 1 and 5: ", np.linspace(1,5,11))
```
