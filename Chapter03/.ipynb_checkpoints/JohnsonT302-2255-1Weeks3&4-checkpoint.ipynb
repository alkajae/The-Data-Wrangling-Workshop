```python
people = ['Anna', 'Brandon', 'Chen', 'David', 'Emily', 'Farook', 'Gagan', 'Hamish', 'Imran', 'Joseph', 'Katherine', 'Lily']
age = [21, 12, 32, 45, 37, 18, 28, 52, 5, 40, 48, 15]
weight = [55, 35, 77, 68, 70, 60, 72, 69, 18, 65, 82, 48]
height = [160, 135, 170, 165, 173, 168, 175, 159, 105, 171, 155, 158]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(age,weight)
plt.show()
```

```python
#adding plot details
#enlarging the figure of the size
plt.figure(figsize=(8,6))
#adding title
plt.title("Plot of Age vs. Weight (in kgs)", fontsize=20)
#adding x and y axis labels
plt.xlabel('Age (years)', fontsize=16)
plt.ylabel('Weight (kgs)', fontsize=16)
#adding a grid
plt.grid(True)
#need to increase the y limit to 100 to give more room to add Weights after 18-20 years of age
plt.ylim(0,100)
#adding ticks 
plt.xticks([i*5 for i in range(12)],fontsize=15)
plt.yticks(fontsize=15)
#changing the color of dots to organge and gives the dots a black outline s is making the size bigger 
plt.scatter(x=age, y=weight, c='orange', s=150, edgecolor='k')
#adding text within the scatterplot
plt.text(x=20, y=85, s="Weights after 18-20 years of age", fontsize=15)
#creates the line to show the weights after 18-20, lw thickens the dashes
plt.vlines(x=20, ymin=0, ymax=80, linestyles='dashed', color='blue', lw=3)
#adds the legend - have to use bracket to show all the words within the legend - parenthases don't work
plt.legend(['Weight in kg'], loc=2,fontsize=12)
plt.show()
```

```python
#fill in the missing values using FILL
superstore_missing.fillna('FILL')
```

```python
#fill in the specific columns with the Fill string
superstore_missing[['Customer','Product']].fillna('FILL')
```

```python
#fill the values for sales using ffill - will take the value before the NA to fill in the next
superstore_missing['Sales'].fillna(method='ffill')
#used 2 to fill in 3
```

```python
#fill using bfill - takes the value after the missing value to fill it
superstore_missing['Sales'].fillna(method='bfill')
```

```python
#dropna syntax
#axis - 0 for dropping rows with missing values, 1 for dropping columns with missing values 
#how - determines if a row or column is removed from a DataFrame when we have at least one NAN value or all NAN values
#thresh - requires many non-NAN values to keep the row/column
#dataframe.dropna(axis,how,thresh,subset,inplace)
```

*Exercise 4.12: Flexibility of User-Defined Functions with the Apply Method*

```python
#create a user-defined function
def categorize_sales(price):
    if price < 50:
        return "Low"
    elif price < 200:
        return "Medium"
    else:
        return "High"
```

```python
superstore_sample_12 = superstore[['Customer Name','State','Sales']].sample(n=100)
```

```python
superstore_sample_12.head(10)
```

```python
#use the apply method to categorize the data based on sales
superstore_sample_12['Sales Price Category']=superstore_sample_12['Sales'].apply(categorize_sales)
superstore_sample_12.head(10)
```
